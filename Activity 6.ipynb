```python
# Activity: First Program in Python with NumPy and Matplotlib

# Part 1: Calculate the maximum height

# Given parameters
v0 = 5          # Initial velocity (m/s)
g = 9.81        # Gravity (m/s^2)
```

```python
# Create 2000 time steps between 0 and 2 seconds
t = np.linspace(0, 2, 2000)
```

```python
# Calculate vertical position
y = v0*t - 0.5*g*t**2
```

```python
# Find maximum height using a for loop
max_y = y[0]
max_t = t[0]

for i in range(1, len(t)):

    if y[i] > max_y:
        max_y = y[i]
        max_t = t[i]

# Print results
print("Maximum height =", max_y, "m")
print("Time at maximum height =", max_t, "s")
```

```python
# Part 2: Calculate the derivative

# Derivative of position with respect to time
dy_dt = v0 - g*t
```

```python
# Plot results
plt.figure(figsize=(10, 8))
```

```python
# First subplot: Vertical position
plt.subplot(211)

plt.plot(t, y, 'k-', label='Vertical position')
plt.plot(max_t, max_y, 'r*', markersize=12,
         label='Maximum height')

plt.xlabel('Time (s)')
plt.ylabel('Height (m)')
plt.title('Vertical Position vs Time')
plt.legend()
plt.grid(True)
# Second subplot: Derivative
plt.subplot(212)

plt.plot(t, dy_dt, 'b-', label='Derivative (Velocity)')

plt.axhline(0, color='k', linestyle='--')

plt.xlabel('Time (s)')
plt.ylabel('Velocity (m/s)')
plt.title('Derivative of Vertical Position')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
